A multiplicação continua em uma função pequena, como no Agente matemático. O modelo chama essa ferramenta, e nós conferimos o valor que ela devolveu. Todas as propostas da loja são fictícias.

## 1. Conectar o modelo

Esta célula importa o LangChain, lê o segredo e configura o Ollama. A primeira pergunta será enviada no passo 3. Não é necessário instalar o Ollama no computador nem baixar o modelo.

````python
import os, json
os.environ['LANGSMITH_TRACING'] = 'false'
os.environ['LANGCHAIN_TRACING_V2'] = 'false'
os.environ['OTEL_SDK_DISABLED'] = 'true'
from dotenv import load_dotenv
load_dotenv('.env')
from langchain.agents import create_agent
from langchain_ollama import ChatOllama
from langchain_core.tools import tool
from langchain_core.messages import ToolMessage
chave = os.environ.get('OLLAMA_API_KEY')
if os.environ.get('GOOGLE_API_KEY'):  # Gemini, plano grátis do Google AI Studio
    from langchain_google_genai import ChatGoogleGenerativeAI
    modelo = ChatGoogleGenerativeAI(model='gemini-3.5-flash-lite', max_output_tokens=256)
elif chave:  # nuvem do Ollama (exige créditos)
    modelo = ChatOllama(model='gemma4:31b', base_url='https://ollama.com',
        temperature=1, num_predict=256, validate_model_on_init=False,
        client_kwargs={'headers': {'Authorization': 'Bearer ' + chave}, 'timeout': 60})
else:  # Ollama local e gratuito: ollama pull qwen2.5:7b
    modelo = ChatOllama(model='qwen2.5:7b', temperature=1, num_predict=256)
print('Modelo configurado. A chave não será exibida.')
````

## 2. Definir a ferramenta, os dados e os papéis

A ferramenta faz a conta em Python. A referência 6×7 continua sendo 42; o subtotal das camisetas é outra conta. Depois criamos os três agentes. Ana recebe a ferramenta; Bruno e Carla trabalham com os pareceres que chegam.

O estoque inicial faz parte do contexto enviado ao modelo. Os agentes escrevem propostas em texto; este exemplo não altera um sistema de estoque.

````python
@tool
def multiplicar(a: int, b: int) -> int:
    """Multiplica dois números e devolve o resultado exato."""
    return a * b

pedido = {'id': 'P001', 'item': 'camiseta', 'quantidade': 2}
preco_unitario = 35
estoque_inicial = 8
referencia = multiplicar.invoke({'a': 6, 'b': 7})
contexto = json.dumps({'pedido': pedido, 'preco_unitario': preco_unitario,
                      'estoque_inicial': estoque_inicial}, ensure_ascii=False)
print('Dados preparados:', contexto)
````

````python
ana = create_agent(modelo, tools=[multiplicar], system_prompt=(
    'Você é Ana, do estoque. Use multiplicar para calcular o subtotal. '
    'Se a quantidade for positiva e couber no estoque, proponha reserva e mostre o restante. '
    'Caso contrário, informe SEM_ESTOQUE. Responda em português, em até três linhas.'))
bruno = create_agent(modelo, tools=[], system_prompt=(
    'Você é Bruno, da logística. O parecer de Ana é uma proposta fictícia, sem confirmação real. '
    'Com reserva proposta, escreva uma proposta de embalar; sem estoque, recuse a liberação. '
    'Não diga reserva confirmada nem afirme execução. Responda em português, em até três linhas.'))
carla = create_agent(modelo, tools=[], system_prompt=(
    'Você é Carla, revisora. Confira a conta, o estoque e a proposta de Bruno. '
    'Confira se alguma proposta foi tratada como confirmação. Aponte erros e lembre o caráter fictício. '
    'Responda em português, em até três linhas.'))
````

## 3. Ana consulta a ferramenta

Preveja o subtotal e o estoque restante. O agente recebe os dados e pode chamar multiplicar. A resposta guarda as mensagens do modelo e da ferramenta.

````python
resposta_ana = ana.invoke({'messages': [{'role': 'user', 'content': contexto}]},
                         config={'recursion_limit': 8})
texto_ana = resposta_ana['messages'][-1].text
print('Ana — estoque:', texto_ana)
````

## 4. Bruno recebe o parecer de Ana

O texto de Ana entra explicitamente na próxima mensagem. Observe se Bruno respeita a situação do estoque.

````python
mensagem_bruno = contexto + '\nParecer de Ana:\n' + texto_ana
resposta_bruno = bruno.invoke({'messages': [{'role': 'user', 'content': mensagem_bruno}]},
                             config={'recursion_limit': 8})
texto_bruno = resposta_bruno['messages'][-1].text
print('Bruno — logística:', texto_bruno)
````

## 5. Carla revisa as duas respostas

Carla recebe o pedido e os dois pareceres. Leia a revisão antes de olhar a conferência em Python: ela identificou os valores e alguma inconsistência?

````python
mensagem_carla = (contexto + '\nParecer de Ana:\n' + texto_ana
                 + '\nParecer de Bruno:\n' + texto_bruno)
resposta_carla = carla.invoke({'messages': [{'role': 'user', 'content': mensagem_carla}]},
                             config={'recursion_limit': 8})
print('Carla — revisão:', resposta_carla['messages'][-1].text)
````

## Conferir o que foi executado

A conferência abaixo usa a mensagem da ferramenta, e não uma conta copiada da resposta do modelo. Para o caso-base, esperamos subtotal 70 e referência 42. Os nomes de modelo nas três etapas mostram a origem das respostas.

Compare também os textos: o estoque esperado é 6, e embalar continua sendo uma proposta. A conferência da multiplicação não prova que todo o texto dos agentes está correto.
Sem estoque suficiente, a conferência exige SEM_ESTOQUE no parecer e permite que Ana dispense a conta. O número de ferramentas observadas fica visível.


````python
ferramentas = [m for m in resposta_ana['messages'] if isinstance(m, ToolMessage)]
assert all(m.status == 'success' for m in ferramentas), 'Confira a ferramenta de Ana.'
if 0 < pedido['quantidade'] <= estoque_inicial:
    assert ferramentas, 'Ana precisa calcular o subtotal com a ferramenta.'
    subtotal_observado = float(ferramentas[-1].content)
    assert subtotal_observado == pedido['quantidade'] * preco_unitario, 'O subtotal não confere.'
    print('Ferramenta:', ferramentas[-1].name, '→', subtotal_observado)
else:
    assert 'SEM_ESTOQUE' in texto_ana.upper(), 'Confira o parecer de estoque.'
    print('Sem estoque suficiente; não exigimos cálculo. Ferramentas:', len(ferramentas))
print('Referência 6×7:', referencia)
for nome, etapa in [('Ana', resposta_ana), ('Bruno', resposta_bruno), ('Carla', resposta_carla)]:
    print(nome, 'usou:', etapa['messages'][-1].response_metadata.get('model') or etapa['messages'][-1].response_metadata.get('model_name'))
````

### Equipe de software: planejar, codificar, revisar e versionar

Arquitetura **sequencial com laço de revisão**: Python define a ordem; o revisor pode devolver o código ao desenvolvedor (no máximo duas rodadas).

- **Paulo (planejamento)** quebra o pedido em passos curtos.
- **Dora (desenvolvimento)** escreve a função em Python.
- **Rui (revisão)** usa `verificar_sintaxe` e responde APROVADO ou REPROVADO com o motivo.
- **Vera (versionamento)** usa `proxima_versao` e propõe a mensagem de commit.

As ferramentas fazem o que precisa ser exato; nenhum código gerado é executado, apenas analisado.

````python
import ast, re

@tool
def verificar_sintaxe(codigo: str) -> str:
    """Confere se o código Python é sintaticamente válido, sem executá-lo."""
    try:
        ast.parse(codigo)
        return 'SINTAXE_OK'
    except SyntaxError as e:
        return f'ERRO_SINTAXE linha {e.lineno}: {e.msg}'

@tool
def proxima_versao(atual: str, tipo: str) -> str:
    """Calcula a próxima versão semântica. tipo deve ser major, minor ou patch."""
    ma, mi, pa = map(int, atual.split('.'))
    return {'major': f'{ma + 1}.0.0', 'minor': f'{ma}.{mi + 1}.0', 'patch': f'{ma}.{mi}.{pa + 1}'}[tipo]

def rodar(agente, texto):
    resposta = agente.invoke({'messages': [{'role': 'user', 'content': texto}]},
                             config={'recursion_limit': 8})
    return resposta['messages'][-1].text, resposta['messages']

def extrair_codigo(texto):
    bloco = re.search(r'```(?:python)?\n(.*?)```', texto, re.S)
    return (bloco.group(1) if bloco else texto).strip()

paulo = create_agent(modelo, tools=[], system_prompt=(
    'Você é Paulo, do planejamento. Quebre o pedido em no máximo quatro passos curtos '
    'para implementar uma única função Python. Não escreva código. Responda em português.'))
dora = create_agent(modelo, tools=[], system_prompt=(
    'Você é Dora, desenvolvedora. Siga o plano e escreva só a função Python pedida, curta, '
    'com docstring, dentro de um bloco ```python. Se houver revisão, corrija o que foi apontado.'))
rui = create_agent(modelo, tools=[verificar_sintaxe], system_prompt=(
    'Você é Rui, revisor. Chame verificar_sintaxe com o código recebido. Confira se ele cumpre o pedido. '
    'Comece com APROVADO ou REPROVADO e explique em até duas linhas, em português.'))
vera = create_agent(modelo, tools=[proxima_versao], system_prompt=(
    'Você é Vera, do versionamento. Funcionalidade nova compatível é minor; correção é patch; '
    'quebra de compatibilidade é major. Chame proxima_versao e escreva uma mensagem de commit '
    'no padrão Conventional Commits. Responda em português, em até três linhas.'))

tarefa = ('Criar a função calcular_total(itens, cupom=None) para a loja. itens é uma lista de '
          'dicionários com preco e quantidade. O cupom DESC10 dá 10% de desconto. '
          'Quantidade negativa deve gerar ValueError.')
versao_atual = '1.2.0'

plano, _ = rodar(paulo, tarefa)
print('Paulo — plano:\n', plano)

revisao = ''
for rodada in range(1, 3):
    pedido_dev = f'Pedido: {tarefa}\nPlano:\n{plano}' + (f'\nRevisão anterior:\n{revisao}' if revisao else '')
    texto_dora, _ = rodar(dora, pedido_dev)
    codigo = extrair_codigo(texto_dora)
    revisao, msgs_rui = rodar(rui, f'Pedido: {tarefa}\nCódigo:\n{codigo}')
    print(f'\nRodada {rodada} — Dora:\n{codigo}\nRui: {revisao}')
    if revisao.strip().upper().startswith('APROVADO'):
        break

texto_vera, msgs_vera = rodar(vera, f'Versão atual: {versao_atual}\nPedido: {tarefa}\n'
                                     f'Revisão: {revisao}\nCódigo:\n{codigo}')
print('\nVera — versionamento:', texto_vera)
````

````python
# Conferência: a verdade vem das ferramentas e do ast, não do texto dos agentes.
sintaxe = [m for m in msgs_rui if isinstance(m, ToolMessage)]
versoes = [m for m in msgs_vera if isinstance(m, ToolMessage)]
assert sintaxe, 'Rui precisa chamar verificar_sintaxe.'
assert versoes and versoes[-1].status == 'success', 'Vera precisa chamar proxima_versao.'
assert verificar_sintaxe.invoke({'codigo': codigo}) == 'SINTAXE_OK', 'O código final tem erro de sintaxe.'
assert 'def calcular_total' in codigo, 'A função pedida não apareceu.'
assert proxima_versao.invoke({'atual': '1.2.0', 'tipo': 'minor'}) == '1.3.0'  # autoteste da ferramenta
print('Rodadas de revisão:', rodada, '| Sintaxe:', sintaxe[-1].content, '| Versão proposta:', versoes[-1].content)
````
